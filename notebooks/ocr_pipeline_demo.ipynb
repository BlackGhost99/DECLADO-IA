# DECLADO-IA · Démo Pipeline OCR

Ce notebook montre comment exécuter le pipeline OCR local (images/PDF) et obtenir un JSON normalisé, ainsi qu'un calcul de CER de base si le *ground truth* est présent (fichier *.gt.txt).

```python
# Imports et configuration
from pathlib import Path
import json

from app.ocr_pipeline.pipeline import run_document

# Exemple: chemin d'entrée
sample = Path('data/raw').glob('*.*')
sample = next(sample, None)
print('Sample found:', sample)

if sample is not None:
    res = run_document(str(sample), lang='eng+fra')
    print('Pages:', res['num_pages'])
    print('CER:', res.get('cer'))
    print('Normalized:', json.dumps(res['normalized'], ensure_ascii=False, indent=2))
else:
    print('Aucun fichier dans data/raw. Placez un PDF/PNG/JPG pour tester.')
```
